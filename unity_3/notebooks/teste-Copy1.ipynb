```python
import wikipedia
from collections import deque
import networkx as nx


# ==========================
# CONFIGURAÇÕES
# ==========================

SEEDS = [
    "Cristiano Ronaldo",   
    "Monte Everest",  
    "Python (programming language)",
    "French Revolution",
    "Albert Einstein"                   
]


MAX_DEPTH = 2
MAX_LINKS_PER_PAGE = 40
STOPS = {
    "Portal", "Category", "Help", "Draft",
    "Template", "Module",
    "File", "Image",
    "Wikipedia", "Talk", "User",
    "MediaWiki",
}


g = nx.DiGraph()

# ==========================
# EXPLORA UMA SEED (BFS)
# ==========================

def explore_seed(start_page):
    queue = deque()
    queue.append((0, start_page))

    visited = set()
    failed = set()

    print(f"\n========== SEED: {start_page} ==========\n")

    while queue:
        layer, page = queue.popleft()

        if layer > MAX_DEPTH:
            continue

        if page in visited:
            continue
        visited.add(page)

        print(f"[Layer {layer}] {page}")

        try:
            wiki = wikipedia.page(page)
        except:
            print("❌ Falhou:", page)
            failed.add(page)
            continue

        if len(wiki.content) < 400:
            continue

        count = 0
        for link in wiki.links:
            if any(link.startswith(stop) for stop in STOPS):
                continue
            if link.lower().startswith("list of"):
                continue
            if count >= MAX_LINKS_PER_PAGE:
                break

            g.add_edge(page, link)

            if link not in visited and link not in failed:
                queue.append((layer + 1, link))

            count += 1

    print(f"Seed finalizada: {start_page}\n")


# ==========================
# PROCESSAR TODAS AS SEEDS
# ==========================

for seed in SEEDS:
    explore_seed(seed)


# ==========================
# LIMPEZA 1 — remover loops
# ==========================
print("\nRemovendo self-loops...")
g.remove_edges_from(nx.selfloop_edges(g))


# ==========================
# LIMPEZA 2 — eliminar duplicatas tipo:
# 'network' e 'networks'
# ==========================

duplicates = [(node, node + "s") for node in g if node + "s" in g]

print("Duplicatas plurais encontradas:", duplicates)

for dup in duplicates:
    g = nx.contracted_nodes(g, *dup, self_loops=False)


# ==========================
# LIMPEZA 3 — remover duplicatas por:
# '-' versus ' ' (hífen vs espaço)
# ==========================

duplicates = []
for node in list(g.nodes()):
    alt = node.replace("-", " ")
    if alt != node and alt in g:
        duplicates.append((node, alt))

print("Duplicatas com hífen/espaço:", duplicates)

for dup in duplicates:
    g = nx.contracted_nodes(g, *dup, self_loops=False)


# ==========================
# LIMPEZA 4 — remover atributo 'contraction'
# (GraphML NÃO aceita dicts)
# ==========================

nx.set_node_attributes(g, 0, "contraction")
nx.set_edge_attributes(g, 0, "contraction")


# ==========================
# SALVAR
# ==========================

nx.write_graphml(g, "graph_aed2.graphml")

print("\n=======================")
print("   PROCESSO TERMINADO  ")
print("=======================\n")
print("Total de nós:", len(g.nodes()))
print("Total de arestas:", len(g.edges()))
print("Arquivo salvo: futebol_brasileiro.graphml")
```
